Implementation of Fundamental Image Processing Operations using Python and OpenCV

Steps
1. Install and import the required libraries, including OpenCV, NumPy, and Matplotlib.


2. Load a color image from the local system and display it using both OpenCV and
Matplotlib.


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load image
image = cv2.imread("/content/sample_data/image panda.jpg")

# Check if image loaded
if image is None:
    print("Image not found")
else:
    print("Image loaded successfully")

# Convert BGR to RGB
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Display image
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

3. Examine the image properties such as dimensions, resolution, number of channels, and
data type.


```python

# Get image properties
height, width, channels = image.shape

print("Dimensions:", image.shape)
print("Width:", width)
print("Height:", height)
print("Resolution:", width, "x", height)
print("Number of Channels:", channels)
print("Data Type:", image.dtype)
```

4. Save the image in different file formats (e.g., JPEG and PNG) and compare the outputs.


```python
cv2.imwrite("panda_output.jpg", image)
cv2.imwrite("panda_output.png", image)
print("Image saved in JPEG and PNG formats")
```

5. Convert the image from RGB/BGR to Grayscale, HSV, and LAB color spaces.


```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
lab = cv2.cvtColor(image, cv2.COLOR_BGR2LAB)
plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale")
plt.axis("off")
plt.subplot(1,3,2)
plt.imshow(cv2.cvtColor(hsv, cv2.COLOR_HSV2RGB))
plt.title("HSV")
plt.axis("off")
plt.subplot(1,3,3)
plt.imshow(cv2.cvtColor(lab, cv2.COLOR_LAB2RGB))
plt.title("LAB")
plt.axis("off")
plt.show()
```

6. Perform basic geometric transformations such as resizing, rotation, and
horizontal/vertical flipping.


```python
# Resize
resized = cv2.resize(image, (400, 600))
# Rotation
h, w = image.shape[:2]
center = (w//2, h//2)
matrix = cv2.getRotationMatrix2D(center, 45, 1)
rotated = cv2.warpAffine(image, matrix, (w, h))
horizontal_flip = cv2.flip(image, 1)
vertical_flip = cv2.flip(image, 0)
plt.figure(figsize=(12,8))
plt.subplot(2,2,1)
plt.imshow(cv2.cvtColor(resized, cv2.COLOR_BGR2RGB))
plt.title("Resized")
plt.axis("off")
plt.subplot(2,2,2)
plt.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))
plt.title("Rotated")
plt.axis("off")
plt.subplot(2,2,3)
plt.imshow(cv2.cvtColor(horizontal_flip, cv2.COLOR_BGR2RGB))
plt.title("Horizontal Flip")
plt.axis("off")
plt.subplot(2,2,4)
plt.imshow(cv2.cvtColor(vertical_flip, cv2.COLOR_BGR2RGB))
plt.title("Vertical Flip")
plt.axis("off")
plt.show()
```

7. Generate the complement (negative) of the image by inverting pixel intensity values.


```python
negative = 255 - image
plt.imshow(cv2.cvtColor(negative, cv2.COLOR_BGR2RGB))
plt.title("Negative Image")
plt.axis("off")
plt.show()
```

8. Crop a Region of Interest (ROI) from the image and analyze its characteristics.


```python
roi = image[400:850, 100:700]
plt.imshow(cv2.cvtColor(roi, cv2.COLOR_BGR2RGB))
plt.title("Region of Interest (ROI)")
plt.axis("off")
plt.show()
print("ROI Shape:", roi.shape)
```

9. Display the original and processed images for comparison and observe the effects of
each operation.


```python
plt.figure(figsize=(15,10))

plt.subplot(2,3,1)
plt.imshow(image_rgb)
plt.title("Original")
plt.axis("off")

plt.subplot(2,3,2)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale")
plt.axis("off")

plt.subplot(2,3,3)
plt.imshow(cv2.cvtColor(resized, cv2.COLOR_BGR2RGB))
plt.title("Resized")
plt.axis("off")

plt.subplot(2,3,4)
plt.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))
plt.title("Rotated")
plt.axis("off")

plt.subplot(2,3,5)
plt.imshow(cv2.cvtColor(horizontal_flip, cv2.COLOR_BGR2RGB))
plt.title("Horizontal Flip")
plt.axis("off")

plt.subplot(2,3,6)
plt.imshow(cv2.cvtColor(negative, cv2.COLOR_BGR2RGB))
plt.title("Negative")
plt.axis("off")

plt.tight_layout()
plt.show()
```
