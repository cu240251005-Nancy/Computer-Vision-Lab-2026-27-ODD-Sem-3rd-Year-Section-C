Steps
1. Import the required libraries, including OpenCV, NumPy, and Matplotlib.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

2. Load a low-contrast grayscale image or convert a color image into grayscale for
processing.

```python
image = cv2.imread('/content/sample_data/cute-red-panda-forest-playing-with-mushrooms.jpg')
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
plt.imshow(gray, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

3. Display the original image and generate its histogram to analyze the distribution of
pixel intensity values.


```python
plt.figure(figsize=(12, 5))
# Display image
plt.subplot(1, 2, 1)
plt.imshow(gray, cmap='gray')
plt.title("Original Grayscale Image")
plt.axis("off")
# Display histogram
plt.subplot(1, 2, 2)
plt.hist(gray.ravel(), 256, [0, 256])
plt.title("Original Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.show()
```

4. Perform contrast stretching to improve the dynamic range of pixel intensities.

```python
# Find minimum and maximum pixel values
min_val = np.min(gray)
max_val = np.max(gray)
# Apply contrast stretching
stretched = ((gray - min_val) / (max_val - min_val) * 255).astype(np.uint8)
# Display result
plt.imshow(stretched, cmap='gray')
plt.title("Contrast Stretched Image")
plt.axis("off")
plt.show()
```

5. Apply Histogram Equalization using OpenCV and compare the enhanced image with
the original image.


```python
equalized = cv2.equalizeHist(gray)
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(gray, cmap='gray')
plt.title("Original Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(equalized, cmap='gray')
plt.title("Histogram Equalized Image")
plt.axis("off")
plt.show()
```

6. Implement Contrast Limited Adaptive Histogram Equalization (CLAHE) and observe
its performance on images with varying illumination.


```python
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
clahe_image = clahe.apply(gray)
plt.imshow(clahe_image, cmap='gray')
plt.title("CLAHE Enhanced Image")
plt.axis("off")
plt.show()
```

7. Plot and compare the histograms of the original, histogram-equalized, and CLAHE
enhanced images.


```python
plt.figure(figsize=(10, 6))
plt.hist(gray.ravel(), 256, [0, 256], alpha=0.5, label="Original")
plt.hist(equalized.ravel(), 256, [0, 256], alpha=0.5, label="Histogram Equalized")
plt.hist(clahe_image.ravel(), 256, [0, 256], alpha=0.5, label="CLAHE")
plt.title("Histogram Comparison")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

8. Analyze the visual improvements in terms of brightness, contrast, and feature visibility.


```python
original_brightness = gray.mean()
equalized_brightness = equalized.mean()
clahe_brightness = clahe_image.mean()
original_contrast = gray.std()
equalized_contrast = equalized.std()
clahe_contrast = clahe_image.std()
print("Visual Analysis")
print("Original Brightness :", round(original_brightness, 2))
print("Equalized Brightness:", round(equalized_brightness, 2))
print("CLAHE Brightness    :", round(clahe_brightness, 2))

print("\nOriginal Contrast :", round(original_contrast, 2))
print("Equalized Contrast:", round(equalized_contrast, 2))
print("CLAHE Contrast    :", round(clahe_contrast, 2))

# Display images for visual comparison
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(gray, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(equalized, cmap='gray')
plt.title("Histogram Equalization")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(clahe_image, cmap='gray')
plt.title("CLAHE")
plt.axis("off")

plt.show()
```

9. Compare the effectiveness of Histogram Equalization and CLAHE for different types
of images.


```python

equalized = cv2.equalizeHist(gray)
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
clahe_image = clahe.apply(gray)
plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)
plt.imshow(gray, cmap='gray')
plt.title("Original Image")
plt.axis("off")
plt.subplot(1, 3, 2)
plt.imshow(equalized, cmap='gray')
plt.title("Histogram Equalization")
plt.axis("off")
plt.subplot(1, 3, 3)
plt.imshow(clahe_image, cmap='gray')
plt.title("CLAHE")
plt.axis("off")
plt.show()
```
